## 4. Univariate Analysis

### Business framing:  

Hotels care about trends like average stay length, customer mix, and pricing. A good EDA starts with understanding single variables and their distribution.

### Do the following:
- Select at least 3 individual variables to explore
- Use plots and summary methods (e.g. info(), describe(), etc)  to describe the distribution (hint: we are only looking at the values of one variable, so think of plots you've used in the past that DON'T compare 2 variables.)
- Focus on what matters from a business standpoint (e.g., pricing, cancellations, guest types)

```python
# Your code for univariate analysis (e.g., plots, value counts) 🔧

# Bar plot of cancellations

sns.countplot(x='is_canceled', data=df)
plt.xlabel('Cancellation Status')
plt.ylabel('Number of Bookings')
plt.title('Booking Cancellations')

plt.xticks([0, 1], ['Not Canceled', 'Canceled'])
plt.show()

#Histogram of lead time

plt.hist(df['lead_time'], bins=30)
plt.xlabel('Lead Time (Days)')
plt.ylabel('Number of Bookings')
plt.title('Distribution of Lead Time')
plt.show()

#Bar chart of hotel types

sns.countplot(x='hotel', data=df)
plt.xlabel('Hotel Type')
plt.ylabel('Number of Bookings')
plt.title('Number of Bookings by Hotel Type')
plt.show()
```

- **Variable 2 – Summary and insights:**  I found that the majority of bookings have a lead time shorter than 50 days, and there seems to be an exponential distribution pattern within the histogram. This tells me that most bookings are planned relatively spontaneously.

## 5. Bivariate Analysis

### Business framing:  

Stakeholders often ask: “What drives cancellations?” or “Do longer stays mean higher revenue?” Bivariate analysis helps you uncover those kinds of relationships.

### Do the following:
- Choose 2 relevant variable pairs (e.g., `lead_time` vs. `is_canceled`, or `adr` vs. `customer_type`)
- Use scatterplots, grouped bar plots, or boxplots to explore the relationships
- Interpret what these relationships could mean for the hotel business

```python
# Your code to analyze variable relationships (e.g., scatterplots, grouped bars) 🔧

# Boxplot of cancelations vs. lead time
sns.boxplot(x='is_canceled', y='lead_time', data=df)
plt.xlabel('Cancellation Status')
plt.ylabel('Lead Time (Days)')
plt.title('Lead Time by Cancellation Status')

plt.xticks([0, 1], ['Not Canceled', 'Canceled'])
plt.show()

# Bar chart of cancelations by hotel type
sns.countplot(x='hotel', hue='is_canceled', data=df)
plt.xlabel('Hotel Type')
plt.ylabel('Number of Bookings')
plt.title('Cancellations by Hotel Type')
plt.legend(title='Cancellation Status', labels=['Not Canceled', 'Canceled'])
plt.show()
```

### In Your Response:
1. Relationship 1 – What did you analyze and what insights did you find?
2. Relationship 2 – What did you analyze and what insights did you find?

### ✍️ Your Response: 🔧
- **Relationship 1:**  In the relationship between lead time and cancellation, I found that canceled bookings typically have a higher lead time than non-canceled bookings. This tells us that the longer the lead time is, the more likely the booking reservation will be canceled.
- **Relationship 2:**  In the relationship between hotel type and cancellations, I found that city hotels, although they have more bookings, have a disproportionate amount of cancellations compared to resort hotels. This tells us that customers are significantly more likely to cancel a city hotel reservation than a resort reservation.
