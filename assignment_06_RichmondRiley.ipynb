## 4. Fill or Fix Values in Key Columns

### Business framing:  

Let’s say your manager wants to see a map of listings with prices and review scores. If key fields are blank, the map won’t work. But not all missing values should be filled the same way.

### Do the following:
- Choose 2 columns with missing values
- Use a strategy to fill or flag those values
  - (e.g., median, “unknown”, forward-fill, or a placeholder)
- Explain what you did and why

### In your response:
1. What two columns did you clean?
2. What method did you use for each, and why?
3. What risks are there in how you filled the data?

````python
# Your code for converting column data types 🔧

#price, replace nulls with median

df['price'] = (
    df['price']
    .replace('[\$,]', '', regex=True)
    .astype(float)
)

print("Missing price values before:", df['price'].isnull().sum())

median_price = df['price'].median()

df['price_was_missing'] = df['price'].isnull()

df['price'].fillna(median_price, inplace=True)

print("Missing price values after:", df['price'].isnull().sum())

#review scores rating, nulls replaced with median

median_rating = df['review_scores_rating'].median()

df['rating_was_missing'] = df['review_scores_rating'].isnull()

df['review_scores_rating'].fillna(median_rating, inplace=True)

print("Missing review_scores_rating after:",
      df['review_scores_rating'].isnull().sum())


````

### ✍️ Your Response: 🔧
1. I cleaned price and review_scores_rating

2. for price I replaced the null values with the median price and created a flag column indicating if the price was missing. I did this because the median is a good representation of a normal listing, less influence of outliers, and the flag column tracks which instances were imputed. For review scores rating I replaced the null values with the median rating and created a flag column indicating if the review rating was missing. I did this because the median is a good representation of the neutral, realistic score and I flagged missing instances because this often means they are new listings.

3. potential bias was introduced because new listings might not actually deserve a median review score, and median pricing might not be accurate for the listing location relative to the rest of the neighborhood. Replacing null values with the median also decreases variability in the dataset which will affect further analysis.

## 6. Remove Duplicate Records

### Business framing:  

If a listing appears twice, it could inflate revenue estimates or confuse users. Airbnb needs each listing to be unique and accurate.

### Do the following:
- Check for rows that are exact duplicates
- If your data has an ID column and each ID is supposed to unique, then make sure there are no duplicate IDs
- Remove duplicates if found

### In your markdown:
1. Did you find duplicates?
2. How did you decide what to drop or keep?
3. Why are duplicates risky for Airbnb teams?


````python
# Add code here 🔧

# Count # of exact duplicate rows
duplicateRows = df.duplicated().sum()

print("Number of exact duplicate rows:", duplicateRows)

df[df.duplicated()]

# Check for duplicate IDs
duplicateIDs = df['id'].duplicated().sum()

print("Number of duplicate listing IDs:", duplicateIDs)

df[df['id'].duplicated(keep=False)].sort_values('id')

#remove exact dupes, keep first instance.

df = df.drop_duplicates()

print("Shape after removing exact duplicates:", df.shape)

df = df.drop_duplicates(subset='id', keep='first')

print("Shape after removing duplicate IDs:", df.shape)


````

### ✍️ Your Response: 🔧 🔧
1. No exact duplicate rows were found, and no duplicate listing IDs were found.

2. For exact duplicates, I removed repeated rows using drop_duplicates(). For duplicate listing IDs, I kept the first occurrence and removed the rest.

3. Duplicates are risky for Airbnb teams because they distort pricing and availability analysis and skew review and host performance metrics, which leads to bad business decisions and less profit.
